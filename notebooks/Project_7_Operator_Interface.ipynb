# Project 7 — Operator Interface

This is the reviewer/operator entry point for the configurable Project 7 decision-support prototype.

**Use:** Runtime → Run all. After setup finishes, use the forms and buttons in the operator interface. No Python glue code or manual audit-file movement is required.

The interface intentionally preserves these boundaries:

- structured solicitation metadata is **operator-confirmed**, not silently extracted from the PDF;
- Project 4 output is **triage only**;
- registered evidence is bounded and may be insufficient;
- the system produces a **nonbinding recommendation**;
- a separate authorized human disposition is required;
- **no autonomous external action** is performed;
- a resumable case bundle can be exported and restored after a Colab runtime loss.


```python
# CELL 1 — SETUP PROJECT 7 OPERATOR INTERFACE

from pathlib import Path
import subprocess
import sys
import shutil

REPO_URL = "https://github.com/icmsol/capstone-project-7-integrated-ai-systems-synthesis.git"
REPO = Path("/content/project7")

if REPO.exists():
    shutil.rmtree(REPO)

print("Cloning current Project 7 main...")
subprocess.run(["git", "clone", "-q", REPO_URL, str(REPO)], check=True)

print("Installing operator-interface dependencies...")
subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q", "-r", str(REPO / "requirements_p5_09.txt")],
    check=True,
)
subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "--index-url",
        "https://download.pytorch.org/whl/cpu",
        "torch>=2.5,<3",
    ],
    check=True,
)

commit = subprocess.run(
    ["git", "rev-parse", "--short", "HEAD"],
    cwd=REPO,
    text=True,
    capture_output=True,
    check=True,
).stdout.strip()

print(f"Repository commit: {commit}")
print("Setup complete.")
```
